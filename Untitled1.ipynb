```python
# ================================================
# Titanic Dataset - Survival Analysis
# Data Science with Python Internship - Task 2
# ================================================

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid")

# ------------------------------------------------
# 1. Load the dataset
# ------------------------------------------------
# Loads directly from seaborn's built-in copy - no download needed
df = sns.load_dataset("titanic")

# --- If you have the Kaggle titanic.csv instead, use this: ---
# df = pd.read_csv("titanic.csv")
# df.rename(columns=str.lower, inplace=True)

print("Shape:", df.shape)
print(df.head())

# ------------------------------------------------
# 2. Explore the data
# ------------------------------------------------
print("\n--- Info ---")
df.info()

print("\n--- Missing values ---")
print(df.isnull().sum().sort_values(ascending=False))

# ------------------------------------------------
# 3. Clean the data
# ------------------------------------------------
df_clean = df.copy()

# Fill missing Age with median
df_clean["age"] = df_clean["age"].fillna(df_clean["age"].median())

# Fill missing embarked/embark_town with mode
if "embarked" in df_clean.columns:
    df_clean["embarked"] = df_clean["embarked"].fillna(df_clean["embarked"].mode()[0])
if "embark_town" in df_clean.columns:
    df_clean["embark_town"] = df_clean["embark_town"].fillna(df_clean["embark_town"].mode()[0])

# Drop 'deck' column (too many missing values)
if "deck" in df_clean.columns:
    df_clean.drop(columns=["deck"], inplace=True)

# Drop duplicate rows
df_clean.drop_duplicates(inplace=True)

# Create age groups for analysis
bins = [0, 12, 18, 35, 60, 100]
labels = ["Child (0-12)", "Teen (13-18)", "Young Adult (19-35)", "Adult (36-60)", "Senior (60+)"]
df_clean["age_group"] = pd.cut(df_clean["age"], bins=bins, labels=labels)

print("\n--- Missing values after cleaning ---")
print(df_clean.isnull().sum())
print(df_clean.head())

# ------------------------------------------------
# 4. Analysis Questions
# ------------------------------------------------
sex_col = "sex" if "sex" in df_clean.columns else "Sex"
pclass_col = "pclass" if "pclass" in df_clean.columns else "Pclass"

# Q1: Who survived more - males or females?
survival_by_sex = df_clean.groupby(sex_col)["survived"].mean().round(3) * 100
print("\n--- Survival Rate by Gender (%) ---")
print(survival_by_sex)

# Q2: Did passenger class affect survival chances?
survival_by_class = df_clean.groupby(pclass_col)["survived"].mean().round(3) * 100
print("\n--- Survival Rate by Passenger Class (%) ---")
print(survival_by_class)

# Q3: What was the survival rate by age group?
survival_by_age_group = df_clean.groupby("age_group")["survived"].mean().round(3) * 100
print("\n--- Survival Rate by Age Group (%) ---")
print(survival_by_age_group)

# ------------------------------------------------
# 5. Visualizations
# ------------------------------------------------

# 5.1 Bar chart: Survival by gender
plt.figure(figsize=(6, 4))
sns.barplot(x=sex_col, y="survived", data=df_clean, palette="Set2")
plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")
plt.ylim(0, 1)
plt.show()

# 5.2 Bar chart: Survival by class
plt.figure(figsize=(6, 4))
sns.barplot(x=pclass_col, y="survived", data=df_clean, palette="Set3")
plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.ylim(0, 1)
plt.show()

# 5.3 Histogram: Passenger ages
plt.figure(figsize=(7, 4))
df_clean["age"].plot(kind="hist", bins=20, color="skyblue", edgecolor="black")
plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()

# 5.4 Bonus: Survival rate by age
plt.figure(figsize=(7, 4))
sns.barplot(x="age_group", y="survived", data=df_clean, palette="coolwarm")
plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")
plt.xticks(rotation=20)
plt.ylim(0, 1)
plt.show()

# 5.5 Bonus: Survival count by class
plt.figure(figsize=(7, 4))
sns.countplot(x=pclass_col, hue="survived", data=df_clean, palette="pastel")
plt.title("Survival Count by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

 Summary
1. Gender: Females survived at a much higher rate than males.
2. Class: Survival rate decreases from 1st -> 2nd -> 3rd class.
3. Age: Children had a higher survival rate than adults and seniors.
4. Combined: 1st-class females had the highest survival rate;
   3rd-class males had the lowest.
